```python
def cal_sum(lst):
    total = 0
    for i in lst:
        total = total + i
    return total

print(cal_sum([8, 2, 3, 0, 7]))
```

```python
def reverse_string(s):
    return s[ ::-1]
print(reverse_string("1234abcd"))
   
```

```python
def cal_fact(n):
   fact =1
   for i in range(1,n+1):
       fact = fact *i
   return fact

print(cal_fact(6))
```
