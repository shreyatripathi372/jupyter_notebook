```python
class Car:
    def __init__(self, model, year, color, for_sale):
        self.model = model
        self.year = year
        self.color = color
        self.for_sale = for_sale

car1 = Car("MUSTANG", 2024, "RED", False)
car2 = Car("Corvette", 2025, "Blue", True)

print(car2.model)
print(car2.year)
print(car2.color)
```

```python
def calc_sum(a, b):
    result = a + b
    print(result)

calc_sum(5, 10)
calc_sum(2, 10)
calc_sum(12, 17) 
```

```python
def calc_avg(a,b,c):
    sum=a+b+c
    avg=sum/3
    print(avg)

calc_avg(98,97,96)
```
